## Linear Regression Normal Form Equation

Linear regression aims to find a linear relationship between input features and target variables. For the given dataset, we have:

- Features matrix $X$ of shape (100, 2)
- Target vector $Y$ of shape (100, 1)
- Weight vector $w$ of shape (2, 1)

The linear model is represented as:
$$Y = Xw$$

To find the optimal weights using the normal equation, we minimize the least squares loss:
$$L(w) = \frac{1}{2n}(Y - Xw)^T(Y - Xw)$$

Taking the derivative with respect to $w$ and setting it to zero:
$$\nabla_w L(w) = -\frac{1}{n}X^T(Y - Xw) = 0$$

Solving for $w$:
$$X^TXw = X^TY$$
$$w = (X^TX)^{-1}X^TY$$

This is the normal equation that gives us the analytical solution for the optimal weight vector in linear regression.

## Get data
- Generate synthetic data with 2 features

```python
X = np.stack((np.arange(-5, 5, 0.1), np.arange(-6, 4, 0.1)), axis=1)
weight_vector = np.array([[5], [1]])

Y = np.dot(X, weight_vector)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter(X[:, 0], X[:, 1], Y, c='r', marker='o')

ax.set_xlabel('X1')
ax.set_ylabel('X2')
ax.set_zlabel('Y')

ax.text(0, 0, 0, "Origin", color='black', fontsize=10)
plt.show()
```

## Implement linear regression in normal form. 

```python
residual = Y - np.dot(X, weight_vector)
least_square_loss = residual.T @ residual * 0.5 / X.shape[0]
print("Residual shape:", residual.shape)
print("Least square loss shape:", least_square_loss.shape)

design_matrix = X.T @ X 
inverse_design_matrix = np.linalg.inv(design_matrix)
weight_vector_from_normal_equation = inverse_design_matrix @ X.T @ Y
print("Weight vector from normal equation weight_vector_from_normal_equation", weight_vector_from_normal_equation.flatten())
```
