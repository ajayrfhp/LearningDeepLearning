```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
normal1 = np.random.normal(loc=10, scale=1, size=(10000, 1))
normal2 = np.random.normal(loc=21, scale=10, size=(10000, 1))


def qcd(x):
    q1, q3 = np.percentile(x, 25), np.percentile(x, 75)
    return (q3 - q1) / (q3 + q1)


cv1 = np.std(normal1) / np.mean(normal1)
cv2 = np.std(normal2) / np.mean(normal2)

qcd1 = qcd(normal1)
qcd2 = qcd(normal2)

plt.hist(normal1, bins=100, color='blue', alpha=0.5, label=f'narrow {cv1:.2f} qcd = {qcd1:.2f}')
plt.hist(normal2, bins=100, color='green', alpha=0.5, label=f'wide {cv2:.2f} qcd ={qcd2:.2f}')
plt.legend(loc='upper right')
plt.show()

```
